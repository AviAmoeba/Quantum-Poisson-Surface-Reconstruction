```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit.circuit.library import QFT



def forcing(x):
    forcing_function = x * (x-1)
    return forcing_function





n_state_qubits = 5 
n_pts = 2 ** n_state_qubits

x = np.linspace(0, 1, n_pts)
f_i    = forcing(x)
f_i[0] = 0 ####




qc = QuantumCircuit(2)
qc.draw(output='mpl')
```

```python
plt.subplot(1, 2, 1) 
plt.title('Forcing') 
plt.plot(x, f_i)
plt.xlim(0, 1)
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.grid()
```
